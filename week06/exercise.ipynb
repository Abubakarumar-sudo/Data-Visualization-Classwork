# Week 06 — Hierarchical Visualization: Global Development

A larger 1,700+ row dataset makes the hierarchy and part-to-whole story more informative than a small categorical table.

## Task 1 — Treemap
Show 2007 GDP by continent and country.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("../data/global_development.csv")
d = df[df.year == 2007].copy()
fig = px.treemap(d, path=["continent","country"], values="gdpPercap",
                 color="lifeExp", color_continuous_scale="Blues",
                 title="Europe combines high incomes with consistently long lives",
                 hover_data={"lifeExp":":.1f"})
fig.show()
```

## Task 2 — Sunburst
Show population as the part-to-whole measure.

```python
fig = px.sunburst(d, path=["continent","country"], values="pop",
                  color="lifeExp", color_continuous_scale="Viridis",
                  title="Population is concentrated in a small set of Asian countries",
                  hover_data={"lifeExp":":.1f"})
fig.show()
```

## Task 3 — Treemap vs bar
Use the top 15 countries by GDP per capita to compare readability.

```python
top = d.nlargest(15, "gdpPercap").sort_values("gdpPercap")
fig = px.bar(top, x="gdpPercap", y="country", orientation="h",
             title="A bar chart makes the top-income ranking easier to compare",
             labels={"gdpPercap":"GDP per capita (USD)","country":""})
fig.show()
```
