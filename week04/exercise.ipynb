# Week 04 — Scatter & Bubble Charts: Wine Chemistry

The UCI wine-recognition dataset contains 178 wines, 13 chemical measurements and three cultivars. It supports meaningful multivariate scatter plots.

## Task 1 — Alcohol vs flavanoids
Compare two chemical properties and use colour only for the wine class.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("../data/wine_quality.csv")
fig = px.scatter(df, x="alcohol", y="flavanoids", color="wine_class",
                 hover_data=["proline"],
                 title="Wine classes separate visibly on alcohol and flavanoids",
                 labels={"alcohol":"Alcohol (%)","flavanoids":"Flavanoids"})
fig.update_traces(marker_size=9, opacity=.75)
fig.update_layout(plot_bgcolor="white", paper_bgcolor="white")
fig.show()
```

## Task 2 — Bubble chart
Use proline concentration as bubble size and highlight class 2.

```python
df["role"] = df["wine_class"].apply(lambda x: "Class 2" if x == "class_2" else "Other")
fig = px.scatter(df, x="alcohol", y="color_intensity", size="proline", color="role",
                 hover_data=["wine_class"],
                 title="Class 2 wines occupy a distinct chemical region",
                 labels={"alcohol":"Alcohol (%)","color_intensity":"Colour intensity","proline":"Proline"},
                 color_discrete_map={"Class 2":"#1f77b4","Other":"#cccccc"},
                 size_max=38)
fig.update_layout(plot_bgcolor="white", paper_bgcolor="white")
fig.show()
```
