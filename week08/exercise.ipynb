# Week 08 — Choropleth Maps: African Economic Geography

A new map dataset derived from Natural Earth country attributes. The map uses ISO-3 codes so the data-to-geometry join is explicit and reproducible.

## Task 1 — Population density proxy
Map estimated population by country with a sequential scale.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("../data/africa_indicators.csv")
fig = px.choropleth(df, locations="iso_a3", locationmode="ISO-3",
                    color="population_m", hover_name="country",
                    color_continuous_scale="Blues",
                    title="Population is concentrated in a small number of African countries",
                    labels={"population_m":"Estimated population (millions)"},
                    projection="natural earth")
fig.update_geos(fitbounds="locations", visible=False)
fig.show()
```

## Task 2 — GDP per capita
Use a sequential scale because higher values represent more of the same quantity.

```python
fig = px.choropleth(df, locations="iso_a3", locationmode="ISO-3",
                    color="gdp_per_capita_usd", hover_name="country",
                    color_continuous_scale="Viridis",
                    title="Economic output per person varies sharply across Africa",
                    labels={"gdp_per_capita_usd":"GDP per capita (USD)"},
                    projection="natural earth")
fig.update_geos(fitbounds="locations", visible=False)
fig.show()
```

### Map design decision
A sequential colour scale is used for both measures because zero is a meaningful lower bound and there is no natural midpoint.
