# Week 05 — Distribution Charts: Restaurant Tips

A compact but richer distribution dataset with bill, tip, party size, day, time and smoker status.

## Task 1 — Bill distributions
Cap the display at the 95th percentile and compare lunch vs dinner.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("../data/restaurant_tips.csv")
p95 = df.total_bill.quantile(.95)
d = df[df.total_bill <= p95]
med = d.groupby("time").total_bill.median()

fig = px.histogram(d, x="total_bill", color="time", nbins=35,
                   barmode="overlay", opacity=.65,
                   title=f"Dinner bills are shifted higher; display capped at the 95th percentile (${p95:.1f})",
                   labels={"total_bill":"Total bill ($)","count":"Tables"})
for t, v in med.items():
    fig.add_vline(x=v, line_dash="dash", annotation_text=f"{t} median ${v:.1f}")
fig.update_layout(plot_bgcolor="white", paper_bgcolor="white")
fig.show()
```

## Task 2 — Box plot by day
Rank days by median bill and show outliers.

```python
order = df.groupby("day").total_bill.median().sort_values(ascending=False).index.tolist()
fig = px.box(df, x="total_bill", y="day", category_orders={"day":order},
             points="outliers",
             title="Saturday has the highest typical restaurant bill",
             labels={"total_bill":"Total bill ($)","day":""})
fig.update_layout(plot_bgcolor="white", paper_bgcolor="white")
fig.show()
```
