# Week 02 — Bar Charts: Iris Measurements

A cleaner replacement dataset with 150 observations and three flower species. The goal is to compare typical measurements without misleading axes.

## Task 1 — Average petal length by species
Use a zero baseline and order species from largest to smallest.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("../data/iris.csv")
summary = (df.groupby("species", as_index=False)["petal_length"]
             .mean()
             .sort_values("petal_length", ascending=False))

fig = px.bar(summary, x="petal_length", y="species", orientation="h",
             text="petal_length",
             title="Virginica has the longest petals by a clear margin",
             labels={"petal_length":"Mean petal length (cm)", "species":""})
fig.update_traces(texttemplate="%{text:.2f} cm", textposition="outside")
fig.update_layout(xaxis_range=[0, summary["petal_length"].max()*1.18],
                  plot_bgcolor="white", paper_bgcolor="white")
fig.show()
```

## Task 2 — Largest species gap
Show the difference between the largest and smallest mean petal lengths.

```python
gap = summary["petal_length"].max() - summary["petal_length"].min()
print(f"Gap between largest and smallest species means: {gap:.2f} cm")

fig = px.bar(summary.sort_values("petal_length"),
             x="petal_length", y="species", orientation="h",
             color="species", text="petal_length",
             title=f"The average petal-length gap is {gap:.2f} cm",
             labels={"petal_length":"Mean petal length (cm)", "species":""})
fig.update_traces(texttemplate="%{text:.2f}", textposition="outside")
fig.update_layout(showlegend=False, xaxis_range=[0, summary["petal_length"].max()*1.18],
                  plot_bgcolor="white", paper_bgcolor="white")
fig.show()
```

### Design note
Both charts use a zero baseline because bar length represents magnitude. Labels reduce the need to estimate values from the axis.
