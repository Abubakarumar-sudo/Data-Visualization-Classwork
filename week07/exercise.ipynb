# Week 07 — Heatmap & Waterfall: Car-share Demand

This dataset contains 249 car-share observations. We use hour-like peak indicators and trip duration to reveal demand structure.

## Task 1 — Heatmap
Create a 2D frequency table from rounded latitude/longitude bins.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("../data/carshare_usage.csv")
df["lat_bin"] = df["centroid_lat"].round(2)
df["lon_bin"] = df["centroid_lon"].round(2)
pivot = df.pivot_table(index="lat_bin", columns="lon_bin", values="car_hours",
                       aggfunc="mean", fill_value=0)
fig = px.imshow(pivot, text_auto=".0f", aspect="auto",
                color_continuous_scale="Blues",
                title="Car-hour intensity clusters in a small number of locations",
                labels={"color":"Mean car-hours"})
fig.show()
```

## Task 2 — Waterfall
Aggregate total car-hours by peak-hour category and finish with the grand total.

```python
import plotly.graph_objects as go

adds = df.groupby("peak_hour")["car_hours"].sum().sort_index()
labels = [str(x) for x in adds.index] + ["Total"]
fig = go.Figure(go.Waterfall(
    x=labels, y=adds.tolist()+[adds.sum()],
    measure=["relative"]*len(adds)+["total"],
    text=[f"{v:,.0f}" for v in adds]+[f"{adds.sum():,.0f}"],
    textposition="outside"
))
fig.update_layout(title="A small number of peak-hour groups account for most car-hours",
                  yaxis_title="Car-hours", plot_bgcolor="white")
fig.show()
```
